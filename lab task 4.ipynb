```python
# LUHN Algorithm implementation
card_number = input("Enter a card number: ").replace(" ", "")
total = 0
is_second = False

for i in range(len(card_number) - 1, -1, -1):
    digit = int(card_number[i])
    
    if is_second:
        digit = digit * 2
        if digit > 9:
            digit = digit - 9
        total += digit
        is_second = False
    else:
        total += digit
        is_second = True

if total % 10 == 0:
    print("Valid card number")
else:
    print("Invalid card number")
```

```python
# Remove punctuations from string
text = input("Enter a string: ")
punctuations = '''!()-[]{};:'"\\,<>./?@#$%^&*_~'''
result = ""

for char in text:
    if char not in punctuations:
        result += char

print("String without punctuations:", result)
```

```python

sentence = input("Enter a sentence: ")
choice = input("Sort by (1) words or (2) letters? ")

if choice == "1":
    words = sentence.split()
    words.sort(key=str.lower)
    result = " ".join(words)
    print("Sorted by words:", result)
    
elif choice == "2":
    letters_only = [char.lower() for char in sentence if char != ' ']
    letters_only.sort()
    result = "".join(letters_only)
    print("Sorted by letters:", result)
    
else:
    print("Invalid choice!")
```
